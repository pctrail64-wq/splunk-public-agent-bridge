```python
# @title 🚀 LivePortrait Dashboard {display-mode: "form"}
import os
import subprocess
from IPython.display import HTML
from google.colab import output

def run_command(cmd, cwd=None):
    process = subprocess.Popen(cmd, shell=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, cwd=cwd, text=True)
    for line in process.stdout:
        print(line, end='')
    process.wait()

# 1. Force fix dependencies for Gradio 4.x compatibility
print("Fixing library versions...")
run_command("pip install -U 'huggingface_hub<0.24.0' 'gradio==4.36.1'")

# 2. Setup Environment
if not os.path.exists('/content/LivePortrait'):
    print("Cloning repository...")
    run_command("git clone -b dev https://github.com/camenduru/LivePortrait /content/LivePortrait")

if not os.path.exists('/content/LivePortrait/pretrained_weights'):
    print("Downloading weights...")
    run_command("git clone https://huggingface.co/camenduru/LivePortrait /content/LivePortrait/pretrained_weights")

# 3. Build Cython Extensions
print("Building Cython extensions...")
cython_path = "/content/LivePortrait/src/utils/dependencies/insightface/thirdparty/face3d/mesh/cython"
run_command("python setup.py build_ext --inplace", cwd=cython_path)

# 4. Launch Gradio with Public URL
print("Launching Gradio App...")
os.chdir('/content/LivePortrait')
# The --share flag generates the public gradio.live URL
!python app.py --share
```
